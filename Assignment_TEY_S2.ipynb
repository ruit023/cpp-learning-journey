```python
#Menu
print("=== COOPMART INVENTORY MENU ===")
print("1. Add Item \n2. Delete Item\n3. Search Item Quantity \n4. Display Inventory \n5. Exit")
menu_function=input("Enter youe choice(1-5): ")

#Creating empty list
items=[]
quantity=[]

#define sort_items() function to sort the items list and quantity list simultaniously
def sort_items(items, quantity):
    n = len(items)
    for i in range(n):
        for j in range(i + 1, n):
            if items[i] > items[j]:
                # swap items
                items[i], items[j] = items[j], items[i]
                # swap quantity
                quantity[i], quantity[j] = quantity[j], quantity[i]



#define add_function() function
def add_function(items, quantity):
    new_item = input("Enter item name to add: ").lower()
    new_quantity = int(input("Enter quantity : "))
    items_lower = [item.lower() for item in items]

    if new_item in items_lower:
        index = items_lower.index(new_item)
        quantity[index] += new_quantity
        print("\nThis item already existed in the items list, new quantity updated automatically.")
    else:
        items.append(new_item)
        quantity.append(new_quantity)
        print("Added", new_item, "with", new_quantity, ".")

    sort_items(items, quantity)
    print("Updated item list:", items)
    print("Updated quantity list:", quantity)




#definite delete_function() function
def delete_function(items, quantity):
    delete_item = input("Enter the item to delete: ").lower()

    items_lower = [item.lower() for item in items]

    if delete_item in items_lower:
        index = items_lower.index(delete_item)
        print(items[index], "has been deleted from inventory.")
        items.pop(index)
        quantity.pop(index)
    else:
        print("This item does not exist in the list")

    print("Updated item list:", items)
    print("Updated quantity list:", quantity)



#define search_function
def search_function(items, quantity):
    search_item = input("Enter the item name to search: ").lower()

    items_lower = [item.lower() for item in items]

    if search_item in items_lower:
        index = items_lower.index(search_item)
        print("Item found:", items[index], "- Quantity:", quantity[index])
    else:
        print("This item does not exist in the list")




going_on=bool(True) #assign initial value for repetition



#repetition
while going_on==True:

  #add function option
  if menu_function=="1":
    add_function(items,quantity)

  #delete function option
  elif menu_function=='2':
    delete_function(items,quantity)

  #Search function option
  elif menu_function=='3':
    search_function(items,quantity)

  #Display Inventory
  elif menu_function=='4':
      print("\nCurrent Inventory")
      for i in range (len(items)):
        print("-",items[i],":", quantity[i])

  #Exit function option
  elif menu_function=='5':
    going_on=False

  #Invalid input statement
  else:
    print("Invalid input\n")

  print("_ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _ _")
  print("\n=== COOPMART INVENTORY MENU ===")
  print("1. Add Item \n2. Delete Item\n3. Search Item Quantity \n4. Display Inventory \n5. Exit")
  menu_function=input("Enter youe choice(1-5): ")



print("Thanks for using our system.")
```
